```python
import os
import pytesseract
from PIL import Image
import pandas as pd

# Set Tesseract path
tesseract_path = r"C:\Users\bala.m\AppData\Local\Programs\Tesseract-OCR\tesseract.exe"
pytesseract.pytesseract.pytesseract_cmd = tesseract_path

# Use PyMuPDF (fitz) to render PDF pages to images (no poppler needed!)
import fitz

pdf_path = "4034-387_Combined_E_00_Report.PDF"
output_dir = "ocr_extracted"
os.makedirs(output_dir, exist_ok=True)

print(f"Converting PDF to images using PyMuPDF and extracting text via OCR...")
try:
    # Open PDF with fitz
    doc = fitz.open(pdf_path)
    
    ocr_text_all = []
    for page_num in range(len(doc)):
        print(f"Processing page {page_num+1}/{len(doc)}...")
        
        # Render page to image
        page = doc[page_num]
        pix = page.get_pixmap(matrix=fitz.Matrix(2, 2))  # 2x zoom for better OCR
        img_path = f"{output_dir}/page_{page_num+1}.png"
        pix.save(img_path)
        
        # Open image and run OCR
        image = Image.open(img_path)
        text = pytesseract.image_to_string(image)
        ocr_text_all.append(text)
        
        # Save page text
        with open(f"{output_dir}/page_{page_num+1}_ocr.txt", "w", encoding="utf-8") as f:
            f.write(text)
        
        print(f"  ✓ OCR completed for page {page_num+1}")
    
    doc.close()
    
    # Save all text to one file
    with open(f"{output_dir}/all_ocr_text.txt", "w", encoding="utf-8") as f:
        f.write("\n\n--- PAGE BREAK ---\n\n".join(ocr_text_all))
    
    print(f"\n✓ OCR extraction complete!")
    print(f"  - Individual pages: {output_dir}/page_X_ocr.txt")
    print(f"  - All combined: {output_dir}/all_ocr_text.txt")
    
    # Show first page preview
    if ocr_text_all:
        print(f"\nFirst page preview:\n{ocr_text_all[0][:500]}\n...")
    
except Exception as e:
    print(f"Error during OCR: {e}")
    import traceback
    traceback.print_exc()
```

```python
%pip install -q easyocr

import fitz  # pymupdf
from PIL import Image
import easyocr
import numpy as np
import io
import os

print("Loading EasyOCR model...")
reader = easyocr.Reader(['en'], gpu=False)

pdf_path = "4034-387_Combined_E_00_Report.PDF"
output_dir = "ocr_extracted_easyocr"
os.makedirs(output_dir, exist_ok=True)

print(f"Converting PDF to images and extracting text via EasyOCR...")
try:
    doc = fitz.open(pdf_path)
    all_texts = []
    
    for page_num in range(len(doc)):
        print(f"Processing page {page_num+1}/{len(doc)}...")
        
        # Render page to image
        page = doc[page_num]
        pix = page.get_pixmap(matrix=fitz.Matrix(2, 2))  # 2x zoom for better OCR
        
        # Convert to PIL Image
        img = Image.open(io.BytesIO(pix.tobytes("png")))
        
        # Convert PIL Image to numpy array for EasyOCR
        img_array = np.array(img)
        
        # Extract text using EasyOCR
        results = reader.readtext(img_array, detail=0)
        text = "\n".join(results)
        all_texts.append(text)
        
        # Save page text
        with open(f"{output_dir}/page_{page_num+1}_ocr.txt", "w", encoding="utf-8") as f:
            f.write(text)
        
        print(f"  ✓ OCR completed for page {page_num+1}")
    
    doc.close()
    
    # Save all text to one file
    with open(f"{output_dir}/all_ocr_text.txt", "w", encoding="utf-8") as f:
        f.write("\n\n--- PAGE BREAK ---\n\n".join(all_texts))
    
    print(f"\n✓ OCR extraction complete!")
    print(f"  - Output folder: {output_dir}/")
    print(f"  - Individual pages: {output_dir}/page_X_ocr.txt")
    print(f"  - All combined: {output_dir}/all_ocr_text.txt")
    
    # Show first page preview
    if all_texts:
        print(f"\nFirst page preview:\n{all_texts[0][:500]}\n...")
    
except Exception as e:
    print(f"Error during OCR: {e}")
    import traceback
    traceback.print_exc()
```
