```python
%pip install -q docling

from docling.document_converter import DocumentConverter
from docling.datamodel.base_models import ConversionStatus
import os

pdf_path = "4034-387_Combined_E_00_Report.PDF"
output_dir = "ocr_extracted_docling"
os.makedirs(output_dir, exist_ok=True)

print("Loading Docling converter...")
converter = DocumentConverter()

print(f"Converting PDF with Docling (includes OCR)...")
result = converter.convert(pdf_path)

if result.status == ConversionStatus.SUCCESS:
    # Extract markdown text
    text = result.document.export_to_markdown()
    
    # Save full text
    with open(f"{output_dir}/all_docling_text.txt", "w", encoding="utf-8") as f:
        f.write(text)
    
    # Save by page (if available)
    for page_num, page in enumerate(result.document.pages, 1):
        page_text = page.export_to_markdown()
        with open(f"{output_dir}/page_{page_num}_docling.txt", "w", encoding="utf-8") as f:
            f.write(page_text)
        print(f"  ✓ Page {page_num} extracted")
    
    print(f"\n✓ Docling extraction complete!")
    print(f"  - Output folder: {output_dir}/")
    print(f"  - All combined: {output_dir}/all_docling_text.txt")
    print(f"  - Individual pages: {output_dir}/page_X_docling.txt")
    print(f"\nFirst 500 characters:\n{text[:500]}\n...")
    
else:
    print(f"Conversion failed with status: {result.status}")
```
